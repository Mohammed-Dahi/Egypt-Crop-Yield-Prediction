```python
import pandas as pd

# Keep only the three agricultural elements
crop = egypt[
    egypt["Element"].isin([
        "Area harvested",
        "Production",
        "Yield"
    ])
].copy()

print("Shape:", crop.shape)

print("\nNumber of items:")
print(crop["Item"].nunique())

print("\nElements:")
print(crop["Element"].value_counts())

print("\nItems:")
print(crop["Item"].unique())

pivot = crop.pivot_table(
    index=["Item", "Year"],
    columns="Element",
    values="Value",
    aggfunc="first"
).reset_index()

pivot.columns.name = None

print("Shape after pivot:", pivot.shape)

print("\nColumns:")
print(pivot.columns.tolist())

print("\nMissing values:")
print(pivot.isna().sum())

display(pivot.head(20))

print("Complete records:")
print(
    pivot[
        ["Area harvested", "Production", "Yield"]
    ].notna().all(axis=1).sum()
)
```

```python
# Number of complete records for each crop
complete = pivot.dropna(
    subset=["Area harvested", "Production", "Yield"]
)

crop_stats = (
    complete.groupby("Item")
    .agg(
        Records=("Year", "count"),
        First_Year=("Year", "min"),
        Last_Year=("Year", "max")
    )
    .sort_values("Records", ascending=False)
)

display(crop_stats.head(30))

# Crops with at least 20 complete years
print(
    crop_stats[crop_stats["Records"] >= 20].shape
)

display(
    crop_stats[crop_stats["Records"] >= 20]
)
```

```python
selected_crops = [
    "Wheat",
    "Rice",
    "Maize (corn)",
    "Sugar cane",
    "Tomatoes"
]

print(
    crop_stats.loc[
        crop_stats.index.intersection(selected_crops)
    ]
)

display(
    pivot[
        pivot["Item"].isin(selected_crops)
    ].sort_values(["Item", "Year"]).head(30)
)
```

```python
df_ml = pivot.copy()

# Sort
df_ml = df_ml.sort_values(["Item", "Year"]).reset_index(drop=True)

# Previous-year features
df_ml["Yield_Lag1"] = df_ml.groupby("Item")["Yield"].shift(1)
df_ml["Yield_Lag2"] = df_ml.groupby("Item")["Yield"].shift(2)

df_ml["Production_Lag1"] = df_ml.groupby("Item")["Production"].shift(1)
df_ml["Production_Lag2"] = df_ml.groupby("Item")["Production"].shift(2)

df_ml["Area_Lag1"] = df_ml.groupby("Item")["Area harvested"].shift(1)
df_ml["Area_Lag2"] = df_ml.groupby("Item")["Area harvested"].shift(2)

# Rolling mean using previous years ONLY
df_ml["Yield_RollingMean_3Y"] = (
    df_ml.groupby("Item")["Yield"]
    .transform(lambda x: x.shift(1).rolling(3).mean())
)

# Growth based ONLY on historical values
df_ml["Area_Growth_Lag1"] = (
    df_ml["Area_Lag1"] / df_ml["Area_Lag2"] - 1
)

df_ml["Production_Growth_Lag1"] = (
    df_ml["Production_Lag1"] / df_ml["Production_Lag2"] - 1
)

print("Shape:", df_ml.shape)

print("\nMissing values:")
print(df_ml.isna().sum())

features = [
    "Yield_Lag1",
    "Yield_Lag2",
    "Production_Lag1",
    "Area_Lag1",
    "Yield_RollingMean_3Y",
    "Area_Growth_Lag1",
    "Production_Growth_Lag1"
]

complete_ml = df_ml[features + ["Yield"]].dropna()

print("\nComplete ML records:", len(complete_ml))

display(complete_ml.head(10))
```

```python
ml_data = df_ml[
    features + ["Item", "Year", "Yield"]
].dropna().copy()

print("Final ML shape:", ml_data.shape)

print("\nNumber of unique crops:")
print(ml_data["Item"].nunique())

print("\nRecords per crop:")
print(ml_data["Item"].value_counts().describe())

print("\nTop 20 crops:")
print(ml_data["Item"].value_counts().head(20))

print("\nYear range:")
print(ml_data["Year"].min(), "to", ml_data["Year"].max())

print("\nRecords per year:")
print(ml_data["Year"].value_counts().sort_index().head(10))

print("\nTarget statistics:")
print(ml_data["Yield"].describe())
```
